```python
import numpy as np
a=np.random.randint(1,15,12)
#will return any 12 number between 1 to 15. Each time on running ,these values will change.
print(a)
print('----------------------------------------')
```

```python
b=a.shape #return number of rows and number of columns.
print(b)
```

```python
print('----------------------------------------')
a=a.reshape(3,4)
#convert 1D array into 2D array and total num should be equal to total elements in 1D array 3*2=12.
print(a)
print('----------------------------------------')
```

```python
a=a.reshape(6,2) #6*2=12
print(a)
print('----------------------------------------')
a=a.reshape(2,6) #2*6=12
print(a)
print('----------------------------------------')
a=a.reshape(1,12)
print(a)
print('----------------------------------------')
print(a.ndim)
a=a.reshape(12) #convert 2D INTO 1D
print(a)
print('----------------------------------------')
```

```python
a=a.reshape(2,2,3) #3D conversion
print(a)
print('----------------------------------------')
print(a.ndim) #shows dimentions of given array.
print('----------------------------------------')
```

```python
a=a.reshape(4,-1)
# convert into 2D array, row would be 4 and column would be 12/4 =3. so -1 will be replaced by 3.
print(a)
print('----------------------------------------')
a=a.reshape(6,-1)
# convert into 2D array, row would be 6 and column would be 12/6 =2. so -1 will be replaced by 2.
print(a)
print('----------------------------------------')
a=a.reshape(2,-1)
print(a)
print('----------------------------------------')
```

```python
a=a.reshape(-1,4)
# convert into 2D array, column would be 4 and row would be 12/4 =3. so -1 will be replaced by 3.
print(a)
print('----------------------------------------')
a=a.reshape(-1,2)
print(a)
print('----------------------------------------')
a=a.reshape(-1,3)
print(a)
```
