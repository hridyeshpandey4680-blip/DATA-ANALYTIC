In Python, data can be stored and manipulated using a variety of data structures, each
suited for different types of operations. Here's a summary of the main built-in data
structures, along with examples of how to use them:


1. Lists – Ordered, Mutable, Allows Duplicates

Use when: You need an ordered collection of items you can change

```python
# Create a list
fruits = ["apple", "banana", "cherry"]
# Add elements
fruits.append("orange")
fruits.insert(1, "mango")
# Remove elements
fruits.remove("banana")
popped = fruits.pop() # removes last item
# Access and modify
print(fruits[0]) # Output: apple
fruits[0] = "grape"
# Loop through list
for fruit in fruits:
 print(fruit)
```

2. Tuples – Ordered, Immutable, Allows Duplicates

Use when: You need a fixed-size collection (e.g., coordinates).

```python
coordinates = (10, 20)
print(coordinates[0]) # Output: 10
# Unpacking
x, y = coordinates
```

3. Sets – Unordered, Mutable, No Duplicates

Use when: You need a collection of unique elements.

```python
numbers = {1, 2, 3, 4, 2} # Duplicate 2 will be removed
# Add and remove
numbers.add(5)
numbers.discard(3)
# Set operations
a = {1, 2, 3}
b = {3, 4, 5}
print(a & b) # Intersection
print(a | b) # Union
```

4. Dictionaries – Key-Value Pairs, Unordered (in versions < 3.7),
Mutable

Use when: You need to associate values with keys.

```python
student = {"name": "Alice", "age": 20, "grade": "A"}
# Access and modify
print(student["name"])
student["age"] = 21
# Add and remove
student["major"] = "Math"
del student["grade"]
# Loop through dictionary
for key, value in student.items():
 print(key, value)
```

5. Queues and Stacks – Using collections.deque
More efficient than using lists for FIFO (queue) or LIFO (stack) operations.
Stack (LIFO)

```python
from collections import deque
stack = deque()
stack.append(1)
stack.append(2)
stack.pop() # Removes 2
```

Queue (FIFO)

```python
queue = deque()
queue.append(1)
queue.append(2)
queue.popleft() # Removes 1
```

6. Custom Classes (Objects) – When you need structured
records

```python
class Person:
 def __init__(self, name, age):
     self.name = name
     self.age = age
p1 = Person("Alice", 30)
print(p1.name)
```
